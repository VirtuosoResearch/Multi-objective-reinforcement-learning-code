```python
import math
import os
import json
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import rc

rc('font',**{'family':'sans-serif','sans-serif':['Helvetica']})
mpl.rcParams['savefig.dpi'] = 1200
mpl.rcParams['text.usetex'] = True  # not really needed

import matplotlib.ticker as mticker

# import data from json
with open('momdp_regrets.json', 'r') as f:
    data = json.load(f)

regrets_moucbvi = data['regrets_ucbvi']
regrets_ens = data['regrets_ens']
regrets_stay = data['regrets_stay']
regrets_uniform = data['regrets_uniform']
regrets_random = data['regrets_random']

plt.figure(figsize=(8, 6))
plt.plot(regrets_moucbvi, label=r'$\mathrm{MO}$-$\mathrm{UCBVI}$', lw=3, color='green', linestyle='--')
plt.plot(regrets_ens, label=r'$\mathrm{ENS}$', lw=3, color='red')
plt.plot(regrets_stay, label=r'$\mathrm{Best~stationary~policy}$', lw=3, color='blue', linestyle=':')
plt.plot(regrets_uniform, label=r'$\mathrm{Uniform}$', lw=3, color='yellow', linestyle='-.')
plt.legend(loc='upper right', fontsize=25, bbox_to_anchor=(0.73, 1.03))

plt.xlabel(r'$\mathrm{Number~of~episodes}$', fontsize=30)
plt.ylabel(r'$\mathrm{Total~regret}$', fontsize=30)

plt.xticks(fontsize=40)
plt.yticks([0, 400, 800, 1200, 1600],fontsize=30)

plt.tight_layout()
plt.savefig('momdp_regrets.pdf', dpi=1200, bbox_inches='tight')
plt.show()
```
