```python

import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import rc

rc('font',**{'family':'sans-serif','sans-serif':['Helvetica']})
mpl.rcParams['savefig.dpi'] = 1200
mpl.rcParams['text.usetex'] = True  # not really needed

import matplotlib.ticker as mticker

attr_mean = np.array([-0.4783, -0.075, 0.0026, 0.0027, 0.0009])
attr_std = np.array([0.012, 0.006, 0.003, 0.001, 0.0006])
gamma = np.array([0.5158, 0.1244, 0.0188, 0.0003, 0.0])

plt.figure(figsize=(6, 5))
plt.scatter(gamma, attr_mean, color='royalblue', alpha=1, s=400)
plt.plot(gamma, attr_mean, color='royalblue', linewidth=4, ls='--', label=r'$\mathrm{Attribution}$')
plt.errorbar(gamma, attr_mean, yerr=attr_std, fmt='none')
plt.xlabel(r'$\Gamma^{obj}$', fontsize=40)
plt.ylabel(r'$\mathrm{Attribution}$', fontsize=40)
plt.xticks([0.1, 0.3, 0.5], fontsize=40)
plt.yticks([-0.5, -0.3, -0.1, 0.1], fontsize=40)
plt.grid(lw=0.4, ls=":")

plt.tight_layout()
title_name = r'$\mathrm{Correlation}=0.98$'
plt.title(title_name, fontsize=36)

plt.show()
plt.savefig("gamma_attr.pdf", format="pdf", dpi=1200)
```
