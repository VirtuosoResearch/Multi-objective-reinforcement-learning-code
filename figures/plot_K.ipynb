```python
import math
import os
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import rc

rc('font',**{'family':'sans-serif','sans-serif':['Helvetica']})
mpl.rcParams['savefig.dpi'] = 1200
mpl.rcParams['text.usetex'] = True  # not really needed

import matplotlib.ticker as mticker

N = [2,3,4,5]
HV = [5.04, 5.17, 5.21, 5.19]
HV_err = [0.16, 0.18, 0.11, 0.12]


fig, ax = plt.subplots(figsize=(6, 5))

plt.scatter(N, HV, color='red', marker='o', alpha=1, s=200)
plt.plot(N, HV, color='red', linewidth=4, label=r"$\mathrm{N}$")

ax.errorbar(N, HV, yerr=HV_err, fmt='none', ecolor='red', capsize=8, elinewidth=2, alpha=0.6)

#plt.xscale('log')  # Hessians vary widely, log scale makes it clearer
plt.xlabel(r"$\mathrm{N}$", fontsize=40)
plt.xticks(N, fontsize=40)
ax.ticklabel_format(style='sci', axis='x', scilimits=(0,0))
ax.xaxis.offsetText.set_fontsize(40)
plt.ylabel(r"$\mathrm{HV}(\times 10^6)$", fontsize=40)
plt.yticks([4.8, 5.1, 5.4], fontsize=40)
#plt.title("Connection between Train Hessian and Datamodels Normalized Error", fontsize=40)
plt.grid(lw=0.4, ls=":")
plt.tight_layout()

plt.savefig("N_halfcheetah.pdf", format="pdf", dpi=1200)
plt.show()
```
